```python
# STEP 3: LOAD DATASETS USING PANDAS

us_df = pd.read_csv("USvideos.csv")
gb_df = pd.read_csv("GBvideos.csv")
in_df = pd.read_csv("INvideos.csv")
categories_df = pd.read_csv("categories.csv")

# Add country labels
us_df["country"] = "US"
gb_df["country"] = "GB"
in_df["country"] = "IN"

# Combine all country datasets
all_videos = pd.concat(
    [us_df, gb_df, in_df],
    ignore_index=True
)

print("US Dataset:", us_df.shape)
print("GB Dataset:", gb_df.shape)
print("India Dataset:", in_df.shape)

print("\nCombined Dataset:", all_videos.shape)
print("Categories Dataset:", categories_df.shape)
```

```python
# STEP 4: EXPLORATORY DATA ANALYSIS

print("First 5 Rows:")
display(all_videos.head())

print("\nDataset Information:")
all_videos.info()

print("\nDescriptive Statistics:")
display(all_videos.describe())

print("\nCountry Distribution:")
print(all_videos["country"].value_counts())
```

```python
# STEP 5: DATA QUALITY CHECK

print("Missing Values in Each Column:")
print(all_videos.isnull().sum())

print("\nTotal Missing Values:")
print(all_videos.isnull().sum().sum())

print("\nDuplicate Records:")
print(all_videos.duplicated().sum())
```

```python
# STEP 6: DATA CLEANING

# Create a separate copy
clean_df = all_videos.copy()

# Remove duplicate records
clean_df = clean_df.drop_duplicates()

# Fill missing numeric values with 0
clean_df["likes"] = clean_df["likes"].fillna(0)
clean_df["comment_count"] = clean_df["comment_count"].fillna(0)

# Fill missing descriptions
clean_df["description"] = clean_df["description"].fillna("No description")

# Check the cleaned dataset
print("Original Dataset Shape:", all_videos.shape)
print("Cleaned Dataset Shape:", clean_df.shape)

print("\nRemaining Missing Values:")
print(clean_df.isnull().sum().sum())

print("\nRemaining Duplicate Records:")
print(clean_df.duplicated().sum())

print("\nData Cleaning Completed Successfully!")
```

```python
# STEP 7: DATE CONVERSION

# Convert trending_date into datetime format
clean_df["trending_date"] = pd.to_datetime(
    clean_df["trending_date"],
    format="%y.%d.%m"
)

# Convert publish_time into datetime format
clean_df["publish_time"] = pd.to_datetime(
    clean_df["publish_time"],
    utc=True
)

# Create a new column for publishing month
clean_df["publish_month"] = (
    clean_df["publish_time"].dt.strftime("%Y-%m")
)

# Display the results
print("Date Conversion Completed!")

display(
    clean_df[
        ["video_id", "publish_time", "trending_date", "publish_month"]
    ].head()
)

print("\nDate Column Information:")
print(clean_df[["publish_time", "trending_date"]].dtypes)
```

```python
# STEP 8: CREATE ENGAGEMENT METRICS

# Total engagement
clean_df["total_engagement"] = (
    clean_df["likes"] + clean_df["comment_count"]
)

# Engagement rate
clean_df["engagement_rate"] = np.where(
    clean_df["views"] > 0,
    (clean_df["total_engagement"] / clean_df["views"]) * 100,
    0
)

# Like ratio
clean_df["like_ratio"] = np.where(
    (clean_df["likes"] + clean_df["dislikes"]) > 0,
    (clean_df["likes"] / (
        clean_df["likes"] + clean_df["dislikes"]
    )) * 100,
    0
)

# Display results
display(
    clean_df[
        ["views", "likes", "dislikes",
         "comment_count", "total_engagement",
         "engagement_rate", "like_ratio"]
    ].head()
)

print("\nAverage Engagement Rate:",
      round(clean_df["engagement_rate"].mean(), 2), "%")

print("Average Like Ratio:",
      round(clean_df["like_ratio"].mean(), 2), "%")
```

```python
# STEP 9: MEAN, MEDIAN AND MODE

columns = ["views", "likes", "comment_count"]

for column in columns:
    print(f"\nStatistics for {column.upper()}")

    print("Mean:", round(clean_df[column].mean(), 2))
    print("Median:", round(clean_df[column].median(), 2))
    print("Mode:", clean_df[column].mode()[0])
```

```python
# STEP 10: OUTLIER DETECTION USING IQR

Q1 = clean_df["views"].quantile(0.25)
Q3 = clean_df["views"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = clean_df[
    (clean_df["views"] < lower_limit) |
    (clean_df["views"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nLower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nNumber of Outliers:", len(outliers))

print("\nPercentage of Outliers:",
      round((len(outliers) / len(clean_df)) * 100, 2), "%")
```

```python
# IQR OUTLIER DETECTION FOR LIKES

Q1_likes = clean_df["likes"].quantile(0.25)
Q3_likes = clean_df["likes"].quantile(0.75)

IQR_likes = Q3_likes - Q1_likes

lower_likes = Q1_likes - 1.5 * IQR_likes
upper_likes = Q3_likes + 1.5 * IQR_likes

likes_outliers = clean_df[
    (clean_df["likes"] < lower_likes) |
    (clean_df["likes"] > upper_likes)
]

print("Likes Q1:", Q1_likes)
print("Likes Q3:", Q3_likes)
print("Likes IQR:", IQR_likes)
print("Lower Limit:", lower_likes)
print("Upper Limit:", upper_likes)
print("Number of Likes Outliers:", len(likes_outliers))

print(
    "Outlier Percentage:",
    round(len(likes_outliers) / len(clean_df) * 100, 2),
    "%"
)
```

```python
# STEP 21: IMPROVED ENGAGEMENT FUNNEL

# Keep only the latest trending record for each video and country
latest_videos = (
    clean_df.sort_values("trending_date")
    .drop_duplicates(
        subset=["video_id", "country"],
        keep="last"
    )
)

# Calculate total engagement metrics
funnel_data = pd.DataFrame({
    "Stage": ["Views", "Likes", "Comments"],
    "Count": [
        latest_videos["views"].sum(),
        latest_videos["likes"].sum(),
        latest_videos["comment_count"].sum()
    ]
})

# Create interactive funnel chart
fig = px.funnel(
    funnel_data,
    x="Count",
    y="Stage",
    title="Video Engagement Funnel - Latest Records"
)

fig.show()

display(funnel_data)
```

### Interpretation – Four-Stage Video Engagement Funnel

The four-stage funnel analyses 15,059 unique video-country records. Initially, all 15,059 records are included. A total of 7,529 records have peak views above the median, representing 50% of the original group. Only 1,506 records exceed the 90th percentile of peak views, representing 10%. Finally, 1,475 records meet both the high-view condition and the minimum requirement of 1,000 comments. The largest drop-off occurs between the first and second stages, with 7,530 records lost. This analysis helps identify videos that achieve both high reach and strong audience interaction.

```python
# STEP 22: VIDEO RETENTION ANALYSIS

# Find first and last trending dates for each video
video_lifespan = (
    clean_df.groupby(["video_id", "country"])
    .agg(
        first_trending=("trending_date", "min"),
        last_trending=("trending_date", "max")
    )
    .reset_index()
)

# Calculate observed lifespan in days (inclusive)
video_lifespan["lifespan_days"] = (
    video_lifespan["last_trending"]
    - video_lifespan["first_trending"]
).dt.days + 1

# Calculate retention percentages
retention_days = [2, 3, 5, 7]

retention_results = []

total_videos = len(video_lifespan)

for day in retention_days:
    retained = (
        video_lifespan["lifespan_days"] >= day
    ).sum()

    retention_rate = (retained / total_videos) * 100

    retention_results.append({
        "Day": day,
        "Retained Videos": retained,
        "Retention Rate (%)": round(retention_rate, 2)
    })

retention_df = pd.DataFrame(retention_results)

display(retention_df)
```

```python
# STEP 23: VIDEO RETENTION CHART

plt.figure(figsize=(9, 5))

plt.plot(
    retention_df["Day"],
    retention_df["Retention Rate (%)"],
    marker="o",
    linewidth=2
)

plt.title("Video Retention Rate Over Time")
plt.xlabel("Number of Days")
plt.ylabel("Retention Rate (%)")

plt.xticks([2, 3, 5, 7])
plt.ylim(0, 100)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# STEP 24: IMPROVED COHORT ANALYSIS

# Use one latest record per video and country
cohort_df = (
    clean_df.sort_values("trending_date")
    .drop_duplicates(
        subset=["video_id", "country"],
        keep="last"
    )
)

# Group videos by publication month
cohort_analysis = (
    cohort_df.groupby("publish_month")
    .agg(
        total_videos=("video_id", "nunique"),
        average_views=("views", "mean"),
        average_likes=("likes", "mean")
    )
    .reset_index()
)

display(cohort_analysis)

# Plot cohort performance
plt.figure(figsize=(12, 6))

plt.plot(
    cohort_analysis["publish_month"],
    cohort_analysis["average_views"],
    marker="o"
)

plt.title("Cohort Analysis: Average Views by Publishing Month")
plt.xlabel("Publishing Month")
plt.ylabel("Average Views")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# TrendPulse – Part 3: Data Visualization and Exploratory Data Analysis (EDA)

The main objective of this part is to perform Exploratory Data Analysis (EDA), identify patterns in video performance, analyse audience engagement, and present findings through static and interactive visualizations.

**Dataset Note:** The datasets used in this project are synthetic reference datasets provided for educational and analytical purposes. The findings represent only the supplied data and should not be considered actual YouTube platform-wide statistics.

## 5. Statistical Analysis

```python
# STEP 25: FOUR-STAGE VIDEO ENGAGEMENT FUNNEL

# Get peak metrics for each unique video and country
video_metrics = (
    clean_df.groupby(["video_id", "country"])
    .agg(
        peak_views=("views", "max"),
        peak_comments=("comment_count", "max")
    )
    .reset_index()
)

# Calculate thresholds
median_views = video_metrics["peak_views"].median()
percentile_90 = video_metrics["peak_views"].quantile(0.90)

# Four funnel stages
stage1 = len(video_metrics)

stage2 = (
    video_metrics["peak_views"] > median_views
).sum()

stage3 = (
    video_metrics["peak_views"] > percentile_90
).sum()

stage4 = (
    (video_metrics["peak_views"] > percentile_90) &
    (video_metrics["peak_comments"] >= 1000)
).sum()

# Create funnel table
funnel_counts = [stage1, stage2, stage3, stage4]

funnel_labels = [
    "All Unique Videos",
    "Above Median Peak Views",
    "Above 90th Percentile Views",
    "Above 90th Percentile + 1000 Comments"
]

funnel_df = pd.DataFrame({
    "Stage": funnel_labels,
    "Video Count": funnel_counts
})

funnel_df["Conversion Rate (%)"] = (
    funnel_df["Video Count"] / stage1 * 100
).round(2)

print("Median Peak Views:", median_views)
print("90th Percentile Peak Views:", percentile_90)

display(funnel_df)

# Identify largest drop-off
dropoffs = [
    funnel_counts[i] - funnel_counts[i + 1]
    for i in range(3)
]

print("Largest Drop-off:",
      funnel_labels[dropoffs.index(max(dropoffs))],
      "to",
      funnel_labels[dropoffs.index(max(dropoffs)) + 1])

print("Videos Lost:", max(dropoffs))
```

```python
# STEP 26: RETENTION COMPARISON BY COUNTRY

# Calculate each video's observed trending lifespan
video_lifespan = (
    clean_df.groupby(["video_id", "country"])
    .agg(
        first_trending=("trending_date", "min"),
        last_trending=("trending_date", "max")
    )
    .reset_index()
)

video_lifespan["lifespan_days"] = (
    video_lifespan["last_trending"]
    - video_lifespan["first_trending"]
).dt.days + 1

# Calculate retention for each country
retention_days = [2, 3, 5, 7]
retention_comparison = []

for country in video_lifespan["country"].unique():

    country_data = video_lifespan[
        video_lifespan["country"] == country
    ]

    total_videos = len(country_data)

    for day in retention_days:

        retained = (
            country_data["lifespan_days"] >= day
        ).sum()

        rate = retained / total_videos * 100

        retention_comparison.append({
            "Country": country,
            "Day": day,
            "Total Videos": total_videos,
            "Retained Videos": retained,
            "Retention Rate (%)": round(rate, 2)
        })

retention_country_df = pd.DataFrame(retention_comparison)

display(
    retention_country_df.pivot(
        index="Day",
        columns="Country",
        values="Retention Rate (%)"
    )
)

# Plot country comparison
plt.figure(figsize=(9, 5))

for country in retention_country_df["Country"].unique():

    country_data = retention_country_df[
        retention_country_df["Country"] == country
    ]

    plt.plot(
        country_data["Day"],
        country_data["Retention Rate (%)"],
        marker="o",
        label=country
    )

plt.title("Video Retention Comparison by Country")
plt.xlabel("Number of Days")
plt.ylabel("Retention Rate (%)")
plt.xticks(retention_days)
plt.ylim(0, 100)
plt.legend(title="Country")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Interpretation – Video Retention Comparison by Country

The line chart compares video retention rates across Great Britain, the United States and India over different days. Great Britain has the highest retention rate at every stage, with 69.12% on Day 2 and 14.34% on Day 7. The United States follows with 64.90% on Day 2 and 10.79% on Day 7. India has the lowest retention, decreasing from 56.91% on Day 2 to 5.63% on Day 7. All three countries show a decline in retention over time. A line chart is suitable because it clearly shows changes and trends across multiple time points.

```python
# STEP 27: COHORT ANALYSIS BY FIRST TRENDING MONTH

cohort_df = (
    clean_df.groupby(["video_id", "country"])
    .agg(
        first_trending_date=("trending_date", "min"),
        peak_views=("views", "max"),
        peak_likes=("likes", "max")
    )
    .reset_index()
)

cohort_df["first_trending_month"] = (
    cohort_df["first_trending_date"].dt.to_period("M")
)

cohort_summary = (
    cohort_df.groupby("first_trending_month")
    .agg(
        total_videos=("video_id", "count"),
        average_views=("peak_views", "mean"),
        average_likes=("peak_likes", "mean")
    )
    .reset_index()
)

display(cohort_summary)

plt.figure(figsize=(10, 5))

plt.plot(
    cohort_summary["first_trending_month"].astype(str),
    cohort_summary["average_views"],
    marker="o"
)

plt.title("Cohort Analysis: Average Peak Views by First Trending Month")
plt.xlabel("First Trending Month")
plt.ylabel("Average Peak Views")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
