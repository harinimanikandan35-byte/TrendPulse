```python
# ==========================================
# TRE​​NDPULSE - PROJECT CONFIGURATION
# ==========================================

import pandas as pd
import numpy as np
from datetime import datetime

# Configuration variables
DATA_PATH = "/content/dataset/"

countries = ["US", "GB", "IN"]

VIEW_THRESHOLD = 1000000

MIN_RECORDS = 40000

# Project start time
start_time = datetime.now()

# Formatted project header
print("=" * 50)
print("       TRE​​NDPULSE DATA ANALYSIS")
print("=" * 50)

print(f"Countries selected: {countries}")
print(f"View threshold: {VIEW_THRESHOLD:,}")
print(f"Minimum records required: {MIN_RECORDS:,}")
print(f"Run started at: {start_time}")

print("=" * 50)
```

```python
# ==========================================
# REUSABLE COUNTRY LOADING FUNCTION
# ==========================================

def load_country(path, country, max_rows=None):

    # Read CSV file
    df = pd.read_csv(path)

    # Add country column
    df["country"] = country

    # Convert DataFrame into list of dictionaries
    records = df.to_dict("records")

    # Apply optional row limit
    if max_rows is not None:
        records = records[:max_rows]

    # Display loaded records
    print(f"{country}: {len(records):,} records loaded")

    return records
```

```python
# ==========================================
# LOAD THREE COUNTRY DATASETS
# ==========================================

all_records = []

for country in countries:

    file_path = DATA_PATH + country + "videos.csv"

    records = load_country(file_path, country)

    all_records.extend(records)

print("\nTOTAL RECORDS:", len(all_records))
```

```python
# ==========================================
# CATEGORY LOOKUP DICTIONARY
# ==========================================

categories = pd.read_csv(DATA_PATH + "categories.csv")

category_names = dict(
    zip(
        categories["category_id"],
        categories["category_name"]
    )
)

print("Category Dictionary:")
print(category_names)
```

```python
# ==========================================
# DATA CLEANING
# ==========================================

numeric_columns = [
    "views",
    "likes",
    "dislikes",
    "comment_count"
]

boolean_columns = [
    "comments_disabled",
    "ratings_disabled",
    "video_error_or_removed"
]

# Track conversion failures
conversion_failures = {
    column: 0 for column in numeric_columns
}

# Clean records
cleaned_records = []

for record in all_records:

    new_record = record.copy()

    # Convert numerical columns
    for column in numeric_columns:

        try:
            value = new_record[column]

            if pd.isna(value):
                raise ValueError("Missing value")

            new_record[column] = int(float(value))

        except (ValueError, TypeError):
            conversion_failures[column] += 1
            new_record[column] = 0

    # Convert Boolean columns
    for column in boolean_columns:

        value = new_record[column]

        if pd.isna(value):
            new_record[column] = False
        else:
            new_record[column] = str(value).lower() in [
                "true", "1", "yes"
            ]

    cleaned_records.append(new_record)

print("Data cleaning completed!")
print("Total cleaned records:", len(cleaned_records))

print("\nConversion failures:")

for column, count in conversion_failures.items():
    print(column, ":", count)
```

```python
# ==========================================
# DATE CONVERSION
# ==========================================

from datetime import datetime

date_failures = {
    "trending_date": 0,
    "publish_time": 0
}

for record in cleaned_records:

    # Convert trending date
    try:
        record["trending_date"] = datetime.strptime(
            str(record["trending_date"]),
            "%y.%d.%m"
        ).date()

    except (ValueError, TypeError):
        date_failures["trending_date"] += 1
        record["trending_date"] = None

    # Convert publishing date
    try:
        record["publish_time"] = datetime.strptime(
            str(record["publish_time"]),
            "%Y-%m-%dT%H:%M:%S.000Z"
        )

    except (ValueError, TypeError):
        date_failures["publish_time"] += 1
        record["publish_time"] = None

print("Date conversion completed!")

print("\nDate conversion failures:")

for column, count in date_failures.items():
    print(column, ":", count)
```

```python
# CREATE TUPLES

video_tuples = []

for record in cleaned_records:

    video = (
        record["video_id"],
        record["trending_date"],
        record["views"],
        record["likes"]
    )

    video_tuples.append(video)

print("First 5 tuples:")
print(video_tuples[:5])

print("\nTuple type:", type(video_tuples[0]))
```

```python
# CREATE COUNTRY-WISE SETS

us_videos = {
    r["video_id"] for r in cleaned_records
    if r["country"] == "US"
}

gb_videos = {
    r["video_id"] for r in cleaned_records
    if r["country"] == "GB"
}

in_videos = {
    r["video_id"] for r in cleaned_records
    if r["country"] == "IN"
}

print("Unique US videos:", len(us_videos))
print("Unique GB videos:", len(gb_videos))
print("Unique IN videos:", len(in_videos))
```

```python
# COMMON VIDEOS

common_us_gb = us_videos.intersection(gb_videos)

common_all = us_videos.intersection(
    gb_videos, in_videos
)

# VIDEOS UNIQUE TO EACH COUNTRY

unique_us = us_videos - gb_videos - in_videos

unique_gb = gb_videos - us_videos - in_videos

unique_in = in_videos - us_videos - gb_videos

print("Videos in US and GB:", len(common_us_gb))
print("Videos in all countries:", len(common_all))

print("US-only videos:", len(unique_us))
print("GB-only videos:", len(unique_gb))
print("India-only videos:", len(unique_in))
```

```python
from collections import Counter

channel_counts = Counter(
    record["channel_title"]
    for record in cleaned_records
)

print("Total unique channels:", len(channel_counts))

print("\nTop 15 channels:")

top_15_channels = channel_counts.most_common(15)

for channel, count in top_15_channels:
    print(channel, ":", count)
```

```python
# NESTED DICTIONARY

country_category_summary = {}

for record in cleaned_records:

    country = record["country"]

    category_id = record["category_id"]

    category_name = category_names.get(
        category_id, "Unknown"
    )

    if country not in country_category_summary:
        country_category_summary[country] = {}

    if category_name not in country_category_summary[country]:
        country_category_summary[country][category_name] = 0

    country_category_summary[country][category_name] += 1


print("COUNTRY-WISE CATEGORY SUMMARY")

for country, categories in country_category_summary.items():

    print("\nCountry:", country)

    for category, count in categories.items():
        print(f"{category}: {count}")
```

```python
# SLICING

top_channels = channel_counts.most_common()

top_15 = top_channels[:15]

print("TOP 15 TRENDING CHANNELS")

for channel, count in top_15:
    print(f"{channel}: {count}")
```

### Python Data Structures – Explanation

1. **List:** Used to store all cleaned video records. Lists are mutable.
2. **Dictionary:** Used to store channel counts and category information as key-value pairs. Dictionaries are mutable.
3. **Tuple:** Used to store video ID, trending date, views and likes. Tuples are immutable.
4. **Set:** Used to identify unique video IDs and compare videos across countries.
5. **Set Intersection:** Used to find videos trending in multiple countries.
6. **Set Difference:** Used to identify videos unique to a particular country.
7. **Nested Dictionary:** Used to organise category counts by country.
8. **Slicing:** Used to display the top 15 trending channels.

These data structures help organise, store and analyse the dataset efficiently.

```python
# TITLE FEATURE ANALYSIS

def title_features(title):

    words = str(title).split()

    uppercase_words = 0

    for word in words:

        if word.isupper() and any(c.isalpha() for c in word):
            uppercase_words += 1

    exclamation_marks = 0
    question_marks = 0

    for character in title:

        if character == "!":
            exclamation_marks += 1

        elif character == "?":
            question_marks += 1

    return {
        "uppercase_words": uppercase_words,
        "exclamation_marks": exclamation_marks,
        "question_marks": question_marks,
        "word_count": len(words)
    }


# Test the function

sample_title = "AMAZING Cricket Match! WATCH NOW?"

print(title_features(sample_title))
```

### Variable Scope and Looping – Explanation

**Variable Scope:**
- Global variables are declared outside functions.
- Local variables are declared inside functions.
- Local variables cannot normally be accessed outside their function.

**Looping:**
- For loop: Repeats a block of code over a sequence.
- While loop: Repeats code while a condition is true.
- Range(): Generates a sequence of numbers.
- Break: Stops a loop immediately.
- Continue: Skips the current iteration and moves to the next iteration.

These concepts help process large datasets and analyse video records efficiently.

```python
# VIDEO CLASSIFICATION FUNCTION

def classify_video(record):

    views = record["views"]
    likes = record["likes"]
    dislikes = record["dislikes"]

    if record["ratings_disabled"]:
        return "Unrated"

    total_ratings = likes + dislikes

    if total_ratings == 0:
        return "Low Engagement"

    like_ratio = likes / total_ratings

    if views >= 1_000_000:

        if like_ratio >= 0.90:
            return "Viral Hit"

        else:
            return "Low Engagement"

    elif views >= 100_000:

        if like_ratio >= 0.70:
            return "Steady Performer"

        elif like_ratio < 0.50:
            return "Divisive"

        else:
            return "Low Engagement"

    elif like_ratio < 0.50:
        return "Divisive"

    else:
        return "Low Engagement"
```

```python
from collections import Counter

classification_counts = Counter(
    classify_video(record)
    for record in cleaned_records
)

print("VIDEO CLASSIFICATION RESULTS")

for label, count in classification_counts.items():

    print(f"{label}: {count:,}")

print("\nTotal classified:", sum(classification_counts.values()))
```

### Hand-Tracing Example

Consider a video with:

- Views: 2,000,000
- Likes: 95,000
- Dislikes: 5,000
- Ratings disabled: False

Step 1: Check whether ratings are disabled. They are not.

Step 2: Calculate like ratio.

Like ratio = 95,000 / (95,000 + 5,000) = 0.95

Step 3: Check views. The video has more than 1 million views.

Step 4: Check like ratio. It is greater than 0.90.

Final classification: Viral Hit.

This demonstrates how nested if/elif/else conditions classify videos using transparent rules.

```python
# REUSABLE FUNCTION FOR TOP RECORDS

def get_top_records(records, column, number=10):

    return sorted(
        records,
        key=lambda record: record[column],
        reverse=True
    )[:number]


top_videos = get_top_records(
    cleaned_records,
    "views",
    10
)

print("TOP 10 VIDEOS")

for video in top_videos:
    print(video["title"], video["views"])
```

### Code Refactoring

I created a reusable function called `get_top_records()` to avoid repeating sorting operations. It accepts the dataset, column name and number of records as parameters.

This improves code reusability and makes the program easier to maintain. The default argument allows the function to return 10 records when no number is specified.
